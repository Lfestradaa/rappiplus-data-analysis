## Introducción

RappiPlus es un servicio de suscripción dentro del ecosistema de Rappi, diseñado para incrementar la frecuencia de compra y el valor generado por usuario.

Sin embargo, el equipo de negocio necesita comprender si el servicio está cumpliendo sus objetivos comerciales y si su operación está generando resultados positivos.

Para ello, se plantea analizar diferentes aspectos del desempeño del servicio a partir de datos de pedidos, catálogo de productos, inversión en marketing y comportamiento de los usuarios.

El análisis busca responder preguntas clave:

- ¿Los usuarios están generando mayores ingresos a través de sus compras?
- ¿El modelo de negocio está generando rentabilidad?
- ¿En qué etapas del proceso de compra se pierden oportunidades de conversión?
- ¿Los usuarios regresan y mantienen su actividad en la plataforma?
- ¿Los cambios implementados en el proceso de compra generan un impacto significativo?

A través de herramientas de análisis de datos, consultas SQL, análisis estadístico y Business Intelligence, se desarrollarán diferentes etapas que permitirán evaluar el desempeño del servicio e identificar oportunidades de mejora.

## 🔹 Paso 2: Analizar si el negocio es rentable

````python
#Ingreso total
revenue = orders['monto_total'].sum()
print(f"Ingreso total: ${revenue:,.2f}")

#Costo total
orders_costos = orders.merge(catalog[['nombre_producto', 'costo_unitario']], on='nombre_producto', how='left')
orders_costos['costo_total_producto'] = (orders_costos['cantidad'] * orders_costos['costo_unitario'])
costo_total = orders_costos['costo_total_producto'].sum()
print(f"Costo total: ${costo_total:,.2f}")

#Inversión en marketing
inver_marketing = marketing['gasto'].sum()
print(f"Inversión en marketing: ${inver_marketing:,.2f}")

#Profit
profit = revenue - costo_total - inver_marketing
print(f"Profit: ${profit:,.2f}")

# Margen de beneficio
margen_profit = (profit/revenue)*100
print(f"Margen de beneficio: {margen_profit:.2f}%")
````

Después de descontar estos dos conceptos, el negocio obtuvo un beneficio de **\$2.836.352,81**, equivalente a un margen de beneficio del **29,88%** sobre los ingresos.

````python
# Explorar tabla events
# =========================
query_events = '''
SELECT *
FROM events
LIMIT 10'''
events = pd.read_sql(query_events, con=engine)
events.head(10)
````

````python
# PARTE 1: Totales del funnel
# ======================

query_totals = '''
WITH cte_first_visit AS (
  SELECT DISTINCT id_usuario
  FROM events
  WHERE nombre_evento = 'first_visit'
),

cte_select_item AS (
  SELECT DISTINCT id_usuario
  FROM events
  WHERE nombre_evento = 'select_item'
),

cte_add_to_cart AS (
  SELECT DISTINCT id_usuario
  FROM events
  WHERE nombre_evento = 'add_to_cart'
),

cte_begin_checkout AS (
  SELECT DISTINCT id_usuario
  FROM events
  WHERE nombre_evento = 'begin_checkout'
),

cte_add_payment_info AS (
  SELECT DISTINCT id_usuario
  FROM events
  WHERE nombre_evento = 'add_payment_info'
),

cte_purchase AS (
  SELECT DISTINCT id_usuario
  FROM events
  WHERE nombre_evento = 'purchase'
)


SELECT
  (SELECT COUNT(*) FROM cte_first_visit)   AS usuarios_first_visit,
  (SELECT COUNT(*) FROM cte_select_item)     AS usuarios_select_item,
  (SELECT COUNT(*) FROM cte_add_to_cart) AS usuarios_cart,
  (SELECT COUNT(*) FROM cte_begin_checkout)   AS usuarios_checkout,
  (SELECT COUNT(*) FROM cte_add_payment_info)   AS usuarios_payment_info,
  (SELECT COUNT(*) FROM cte_purchase)   AS usuarios_fpurchase
'''

totals = pd.read_sql(query_totals, con=engine)
totals
````

````python
# PARTE 2: Conversiones
# ======================

query_conversion = '''
WITH cte_first_visit AS (
  SELECT DISTINCT id_usuario
  FROM events
  WHERE nombre_evento = 'first_visit'
),

cte_select_item AS (
  SELECT DISTINCT id_usuario
  FROM events
  WHERE nombre_evento = 'select_item'
),

cte_add_to_cart AS (
  SELECT DISTINCT id_usuario
  FROM events
  WHERE nombre_evento = 'add_to_cart'
),

cte_begin_checkout AS (
  SELECT DISTINCT id_usuario
  FROM events
  WHERE nombre_evento = 'begin_checkout'
),

cte_add_payment_info AS (
  SELECT DISTINCT id_usuario
  FROM events
  WHERE nombre_evento = 'add_payment_info'
),

cte_purchase AS (
  SELECT DISTINCT id_usuario
  FROM events
  WHERE nombre_evento = 'purchase'
)


SELECT
  (SELECT COUNT(*) FROM cte_first_visit)   AS usuarios_first_visit,
  (SELECT COUNT(*) FROM cte_select_item)     AS usuarios_select_item,
  (SELECT COUNT(*) FROM cte_add_to_cart) AS usuarios_cart,
  (SELECT COUNT(*) FROM cte_begin_checkout)   AS usuarios_checkout,
  (SELECT COUNT(*) FROM cte_add_payment_info)   AS usuarios_payment_inf,
  (SELECT COUNT(*) FROM cte_purchase)   AS usuarios_purchase,
  
  -- El calculo de drop off
  
    ROUND(((SELECT COUNT(*) FROM cte_first_visit) - (SELECT COUNT(*) FROM cte_select_item))*100.0
    / NULLIF((SELECT COUNT(*) FROM cte_first_visit), 0),2) AS dropoff_after_first_visit,

    ROUND(((SELECT COUNT(*) FROM cte_select_item) - (SELECT COUNT(*) FROM cte_add_to_cart))*100.0 
    / NULLIF((SELECT COUNT(*) FROM cte_select_item), 0),2) AS dropoff_after_select_item,
   
    ROUND(((SELECT COUNT(*) FROM cte_add_to_cart) - (SELECT COUNT(*) FROM cte_begin_checkout))*100.0
    / NULLIF((SELECT COUNT(*) FROM cte_add_to_cart), 0),2) AS dropoff_after_add_to_cart,
   
    ROUND(((SELECT COUNT(*) FROM cte_begin_checkout) - (SELECT COUNT(*) FROM cte_add_payment_info)) *100.0
    / NULLIF((SELECT COUNT(*) FROM cte_begin_checkout), 0),2) AS dropoff_after_begin_checkout,

    ROUND(((SELECT COUNT(*) FROM cte_add_payment_info) - (SELECT COUNT(*) FROM cte_purchase))*100.0
    / NULLIF((SELECT COUNT(*) FROM cte_add_payment_info), 0),2) AS dropoff_after_payment_info,
    
        -- Usuarios perdidos entre etapas
        
    (SELECT COUNT(*) FROM cte_first_visit) -
    (SELECT COUNT(*) FROM cte_select_item) AS lost_after_first_visit,
    (SELECT COUNT(*) FROM cte_select_item) -
    (SELECT COUNT(*) FROM cte_add_to_cart) AS lost_after_select_item,
    (SELECT COUNT(*) FROM cte_add_to_cart) -
    (SELECT COUNT(*) FROM cte_begin_checkout) AS lost_after_add_to_cart,
    (SELECT COUNT(*) FROM cte_begin_checkout) -
    (SELECT COUNT(*) FROM cte_add_payment_info) AS lost_after_begin_checkout,
    (SELECT COUNT(*) FROM cte_add_payment_info) -
    (SELECT COUNT(*) FROM cte_purchase)  AS lost_after_add_payment_info,
    
      -- Conversión final
    ROUND(100.0*(SELECT COUNT(*) FROM cte_purchase) /
    NULLIF((SELECT COUNT(*) FROM cte_first_visit), 0), 2) AS conversion_final
'''

conversion = pd.read_sql(query_conversion, con=engine)
conversion
````

#### Análisis del comportamiento de los usuarios y conversión

El análisis del comportamiento de los usuarios permitió identificar las diferentes etapas del proceso de compra, desde la primera visita hasta la adquisición de un producto.

Se registraron **7.796 usuarios** en el evento `first_visit` y **6.240 usuarios** en `purchase`, lo que representa una conversión final calculada del **80,04%**.

````python
# Explorar tabla user_activity
# =========================
query_user_activity = '''
SELECT
    COUNT(DISTINCT CASE
        WHEN ua.dias_despues_registro = 7 THEN ua.id_usuario
    END) AS usuarios_con_dia_7,

    COUNT(DISTINCT CASE
        WHEN ua.dias_despues_registro = 14 THEN ua.id_usuario
    END) AS usuarios_con_dia_14,

    COUNT(DISTINCT CASE
        WHEN ua.dias_despues_registro = 21 THEN ua.id_usuario
    END) AS usuarios_con_dia_21
FROM user_activity AS ua;'''
user_activity = pd.read_sql(query_user_activity, con=engine)
user_activity.head(10)
````

````python
# Retención por cohortes
# ======================

query_cohort_retention_final = '''

WITH cohortes AS (
    SELECT
        u.id_usuario,
        DATE_TRUNC('month', MIN(CAST(u.fecha_registro AS DATE))) AS cohorte_mensual
    FROM users AS u
    GROUP BY u.id_usuario
),
retencion AS (
    SELECT
        c.cohorte_mensual,
        COUNT (DISTINCT c.id_usuario) AS clientes_iniciales,
        COUNT(DISTINCT CASE WHEN ua.dias_despues_registro = 7 AND ua.activo = 1 THEN ua.id_usuario END) AS retenido_w1,
        COUNT(DISTINCT CASE WHEN ua.dias_despues_registro = 14 AND ua.activo = 1 THEN ua.id_usuario END) AS retenido_w2,
        COUNT(DISTINCT CASE WHEN ua.dias_despues_registro = 21 AND ua.activo = 1 THEN ua.id_usuario END) AS retenido_w3
    FROM cohortes AS c
    LEFT JOIN user_activity AS ua
        ON c.id_usuario = ua.id_usuario
    GROUP BY c.cohorte_mensual
)

    SELECT
        TO_CHAR(cohorte_mensual, 'YYYY-MM') AS cohorte,
        clientes_iniciales,
        retenido_w1,
        retenido_w2,
        retenido_w3,
        ROUND(retenido_w1::numeric / NULLIF(clientes_iniciales, 0) * 100, 2) AS semana_1,
        ROUND(retenido_w2::numeric / NULLIF(clientes_iniciales, 0) * 100, 2) AS semana_2,
        ROUND(retenido_w3::numeric / NULLIF(clientes_iniciales, 0) * 100, 2) AS semana_3
    FROM retencion
    ORDER BY cohorte_mensual

'''
# Ejecutar la consulta
cohorte_final = pd.read_sql(query_cohort_retention_final, con=engine)
cohorte_final
````

### Metodología de cohortes

1. Se identifica la cohorte de cada usuario según el **mes de registro**.


2. Se calcula la retención semanal: cuántos usuarios **se mantienen activos** en cada semana desde su registro.
   - `retenido_w1`: usuarios activos en la semana 1  
   - `retenido_w2`: usuarios activos en la semana 2  
   - `retenido_w3`: usuarios activos en la semana 3  


3. Se calcula el porcentaje de retención para cada semana, dividiendo los usuarios retenidos entre los clientes iniciales de la cohorte:  
   - `semana_1`: porcentaje de usuarios retenidos en la semana 1  
   - `semana_2`: porcentaje de usuarios retenidos en la semana 2  
   - `semana_3`: porcentaje de usuarios retenidos en la semana 3  

Se revisa que la columna de fecha esté en formato correcto (`DATE`).  
Se realiza la conversión usando: `CAST(fecha_registro AS DATE)`

### Análisis de retención por cohortes

El análisis de cohortes permitió evaluar el comportamiento de los usuarios durante las primeras tres semanas posteriores a su registro.

Los resultados muestran niveles de retención relativamente estables entre las cohortes analizadas:

- **Semana 1:** aproximadamente entre 41,20% y 42,84%.
- **Semana 2:** aproximadamente entre 40,07% y 43,40%.
- **Semana 3:** aproximadamente entre 40,32% y 43,98%.

Esto indica que una proporción importante de los usuarios continúa activa durante las primeras semanas posteriores a su registro. Sin embargo, los resultados no muestran una mejora sostenida entre cohortes ni una disminución uniforme de la retención a medida que transcurren las semanas.

**Comportamiento irregular de las cohortes:** Se identificó un comportamiento que no corresponde al patrón esperado en un análisis de retención. Mientras que la cohorte de enero presenta una disminución progresiva, aunque moderada, en la cantidad de usuarios activos a medida que transcurren las semanas, las demás cohortes muestran fluctuaciones, con incrementos y disminuciones en sus niveles de retención. 

Este comportamiento irregular dificulta identificar una tendencia consistente de abandono de usuarios y puede estar relacionado con diferencias en el comportamiento de las cohortes, la metodología de cálculo o la forma en que se registran los eventos de actividad. Por lo tanto, sería conveniente revisar estos aspectos antes de establecer conclusiones definitivas sobre la retención de usuarios.

### Medidas DAX utilizadas

Se desarrollaron medidas para calcular los indicadores financieros y comerciales utilizados en las visualizaciones.

#### Indicadores comerciales

**Ingresos Totales:** calcula la suma de los ingresos generados por los pedidos.

```dax
Ingresos Totales =
SUM(orders_clean[monto_total])
```

**Cantidad de Órdenes:** contabiliza los pedidos únicos.

```dax
Cantidad de Ordenes =
DISTINCTCOUNT(orders_clean[id_pedido])
```

**Ticket Promedio:** calcula el ingreso promedio por pedido.

```dax
Ticket Promedio =
DIVIDE([Ingresos Totales], [Cantidad de Ordenes], 0)
```

**Cantidad de Productos Vendidos:** suma las unidades comercializadas.

```dax
Cantidad Productos Vendidos =
SUM(orders_clean[cantidad])
```

**Promedio de Productos por Orden:** calcula las unidades promedio por pedido.

```dax
Promedio Productos por Orden =
DIVIDE(
    [Cantidad Productos Vendidos],
    [Cantidad de Ordenes],
    0
)
```

#### Indicadores financieros

**Costo Total de Productos:** calcula el costo de las unidades vendidas, utilizando el costo unitario del catálogo.

```dax
Costo Total Productos =
SUMX(
    orders_clean,
    orders_clean[cantidad] *
    RELATED(catalog_clean[costo_unitario])
)
```

**Gasto Total de Marketing:** suma la inversión realizada en marketing.

```dax
Gasto Total Marketing =
SUM(marketing_clean[gasto])
```

**Profit Total:** calcula el beneficio después de descontar los costos de productos y la inversión en marketing.

```dax
Profit Total =
[Ingresos Totales]
    - [Costo Total Productos]
    - [Gasto Total Marketing]
```

**Margen de Beneficio:** representa la proporción del beneficio calculado sobre los ingresos.

```dax
Margen de Beneficio =
DIVIDE(
    [Profit Total],
    [Ingresos Totales],
    0
)
```

**Profit Bruto Productos:** calcula la diferencia entre los ingresos y los costos de productos, sin descontar marketing.

```dax
Profit Bruto Productos =
[Ingresos Totales] - [Costo Total Productos]
```
